```python
#are older customer more likely to churn compared to younger customers
result.groupby("num_years_antig").mean(numeric_only=True)
```

```python
df3 = result[['num_years_antig', "churn"]]
```

```python
df3 = df3.groupby("num_years_antig")["churn"].mean(numeric_only=True)
```

```python
df3.plot(kind="line", marker="o")
plt.xlabel("Years as Customer")
plt.ylabel("Churn probability")
plt.title("Churn probability by Years as Customer")
plt.grid("True")
plt.show()
#negative correlation with churning, so the smaller amount of years they have been a customer, the more liekely to churn.
```
